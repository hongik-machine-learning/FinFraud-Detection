```python
# === Downstream classifier (sklearn) on encoder features ===
from sklearn.metrics import roc_auc_score, f1_score, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier

# 간단한 로지스틱 회귀
clf_lr = LogisticRegression(max_iter=2000, n_jobs=None, class_weight="balanced")
clf_lr.fit(Z_train, y_train)
pred_proba = clf_lr.predict_proba(Z_test)[:,1]
pred_label = (pred_proba >= 0.5).astype(int)

auc = roc_auc_score(y_test, pred_proba)
f1 = f1_score(y_test, pred_label)
acc = accuracy_score(y_test, pred_label)
print(f"[LR on SSL features] AUC={auc:.4f} | F1={f1:.4f} | Acc={acc:.4f}")

# 선택적으로 MLP 분류기
clf_mlp = MLPClassifier(hidden_layer_sizes=(128,64), max_iter=200, random_state=42)
clf_mlp.fit(Z_train, y_train)
pred_proba_m = clf_mlp.predict_proba(Z_test)[:,1]
pred_label_m = (pred_proba_m >= 0.5).astype(int)

auc_m = roc_auc_score(y_test, pred_proba_m)
f1_m = f1_score(y_test, pred_label_m)
acc_m = accuracy_score(y_test, pred_label_m)
print(f"[MLP on SSL features] AUC={auc_m:.4f} | F1={f1_m:.4f} | Acc={acc_m:.4f}")
```
